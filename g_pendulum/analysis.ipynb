```python
import numpy as np
```

```python
total_time = np.array([213.07, 258.45, 224.57])
total_time_err = np.array([0.01, 0.01, 0.01])
num_period = np.array([75,91,79])
length = 200.6/100
length_err = 0.3/100
```

```python
period = total_time / num_period
period_err = total_time_err / np.sqrt(num_period)

g = 4 * np.pi**2 * length / period**2
g_err = g * np.sqrt((length_err/length)**2 + (2*period_err/period)**2)
```
