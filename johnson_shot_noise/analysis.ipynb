```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
filepath = "data/10-07-noise-vs-temp.csv"
df_full = pd.read_csv(filepath)
df_full.head(5)
```

```python
df_full['Uncertainty (C)'][0]
```

```python
v2_values = np.array([])
v2_uncs = np.array([])
temps = np.array([])
temp_uncs = np.array([])

for temp in df_full['Temp (C)'].unique():
    df_temp = df_full[df_full['Temp (C)'] == temp]
    np.append(temps, temp)
    temp_unc = df_temp['Uncertainty (C)'].iloc[0]
    np.append(temp_uncs, temp_unc)

    vr_values = df_temp[df_temp['V_R (mV)']>0]['V_R (mV)']
    vr_mean = np.mean(vr_values)
    vr_unc = np.std(vr_values, ddof=1)/len(vr_values)**0.5

    vs_values = df_temp[df_temp['V_S (mV)']>0]['V_S (mV)']
    vs_mean = np.mean(vs_values)
    vs_unc = np.std(vs_values, ddof=1)/len(vs_values)**0.5

    # print(f"temp = {temp}")
    # print(f"V_r = {vr_mean} $\pm$ {vr_unc}")
    # print(f"V_s = {vs_mean} $\pm$ {vs_unc}")

    v2 = vr_mean**2 - vs_mean**2
    v2_unc = 2 * ( (vr_mean*vr_unc)**2 + (vs_mean*vs_unc)**2 )**0.5

    np.append(v2_values, v2)
    np.append(v2_uncs, v2_unc)
```
