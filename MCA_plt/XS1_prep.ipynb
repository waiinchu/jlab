```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("uranium.spe", skiprows = 12, skipfooter = 17, engine='python', header=None)
```

```python
plt.hist(df[0], bins = 1000)

plt.title("Natural Uranium")
plt.ylabel("Frequency (counts)")
plt.grid(True, alpha=0.3, linewidth = 0.5)
plt.yscale('log')
plt.savefig("uranium.pdf", dpi=300, bbox_inches="tight")

plt.show()
```
